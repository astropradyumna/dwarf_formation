```python
def get_grsubh_id(snap):
	'''
	This function returns all the SubFindIDs of the group at a given snapshot
	snap: input snapshot
	'''
	central_firstsub  = central_tree['GroupFirstSub'][np.where(central_snaps == snap)]
	central_nsubs = central_tree['GroupNsubs'][snap == central_snaps]
	subh_ids = range(central_firstsub[0], central_firstsub[0] + central_nsubs[0])
	#This zero is just to convert the sequence to a scalar, nothing to do with the group number
	return np.array(subh_ids, dtype = int)
```
