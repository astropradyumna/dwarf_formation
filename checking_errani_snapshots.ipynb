```python
import pandas as pd
import sys 
import matplotlib.pyplot as plt
import numpy as np 


sys.path.append('/bigdata/saleslab/psadh003/misc_files/for_prady')

```

```python
# Read the file 'stars050-004.dat' using pandas
# Assuming columns are separated by single whitespace

path = '/bigdata/saleslab/psadh003/misc_files/for_prady/'
data_050 = pd.read_csv(path + 'stars050-013.dat', delim_whitespace=True, comment='#', header=None)

# Column 3 is fdm = Mmx/Mmx0
fdm_050 = data_050[2].values

# Column 12 is Rhbyrmx0
Rhbyrmx0_050 = data_050[11].values

# Column 14 is L
L_050 = data_050[13].values


# Lets now import another file 'stars025-009.dat'
data_025 = pd.read_csv(path + 'stars025-013.dat', delim_whitespace=True, comment='#', header=None)

# Column 3 is fdm = Mmx/Mmx0
fdm_025 = data_025[2].values

# Column 12 is Rhbyrmx0
Rhbyrmx0_025 = data_025[11].values

# Column 14 is L
L_025 = data_025[13].values

# Let now import another file 'stars012-009.dat'
data_012 = pd.read_csv(path + 'stars012-013.dat', delim_whitespace=True, comment='#', header=None)

# Column 3 is fdm = Mmx/Mmx0
fdm_012 = data_012[2].values

# Column 12 is Rhbyrmx0
Rhbyrmx0_012 = data_012[11].values

# Column 14 is L
L_012 = data_012[13].values

# Let now import another file 'stars006-009.dat'
data_006 = pd.read_csv(path + 'stars006-013.dat', delim_whitespace=True, comment='#', header=None)


# Column 3 is fdm = Mmx/Mmx0
fdm_006 = data_006[2].values

# Column 12 is Rhbyrmx0
Rhbyrmx0_006 = data_006[11].values

# Column 14 is L
L_006 = data_006[13].values
```

```python
fig, ax = plt.subplots(figsize = (6, 6))
ax.plot(np.log10(fdm_050), np.log10(Rhbyrmx0_050), c='red', alpha=0.5)
ax.plot(np.log10(fdm_025), np.log10(Rhbyrmx0_025), c='orange', alpha=0.5)
ax.plot(np.log10(fdm_012), np.log10(Rhbyrmx0_012), c='skyblue', alpha=0.5)
ax.plot(np.log10(fdm_006), np.log10(Rhbyrmx0_006), c='darkblue', alpha=0.5)
ax.axhline(y = np.log10(0.5), color = 'red')
ax.axhline(y = np.log10(0.25), color = 'orange')
ax.axhline(y = np.log10(0.125), color = 'skyblue')
ax.axhline(y = np.log10(0.0625), color = 'darkblue')
ax.set_xlabel('log fdm (Mmx/Mmx0)')
ax.set_ylabel('log Rhbyrmx0')
plt.tight_layout()
# plt.legend()
plt.show()
```
